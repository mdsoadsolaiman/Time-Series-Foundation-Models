```python
import matplotlib.pyplot as plt

# Rank the current full ten-model composite directly from the v2 component file.
score_column = 'Exploratory Composite — Missing Evidence Penalised'
trust_ranking = trust.set_index('Model')[score_column].sort_values()
ax = trust_ranking.plot(kind='barh', figsize=(10, 6), color='#0072B2')
ax.set(title='Current Bitcoin Trust Score ranking (10 models)', xlabel='Exploratory composite score', ylabel='Model', xlim=(0, 100))
ax.grid(axis='x', alpha=0.25)
ax.figure.tight_layout()
plt.show()
```

```python
# Load the frozen transparency and auditability rubric.
rubric = pd.read_csv(ROOT / 'results' / 'bitcoin_transparency_auditability_rubric.csv')

# Display the unchanged rubric.
rubric
```

```python
# Load the frozen Trust Score sensitivity analysis.
sensitivity = pd.read_csv(ROOT / 'results' / 'bitcoin_trust_score_sensitivity_v2.csv')

# Display the unchanged scenario-rank table.
sensitivity.pivot(index='Model', columns='Scenario', values='Rank').sort_values('Base')
```

```python
# Pivot the already-loaded sensitivity ranks for scenario comparison.
rank_sensitivity = sensitivity.pivot(index='Model', columns='Scenario', values='Rank').sort_values('Base')
scenario_order = ['Base', 'Higher transparency', 'Lower transparency']
available_order = [scenario for scenario in scenario_order if scenario in rank_sensitivity.columns]

# Plot lower (better) ranks at the top of the axis.
ax = rank_sensitivity[available_order].plot(kind='bar', figsize=(12, 5), width=0.8)
ax.set(title='Trust Score rank sensitivity to transparency weighting', xlabel='Model', ylabel='Rank')
ax.invert_yaxis()
ax.tick_params(axis='x', rotation=25)
ax.grid(axis='y', alpha=0.25)
ax.legend(title='Scenario', frameon=False)
ax.figure.tight_layout()
plt.show()
```

## Interpretation guardrails
Dimensions are correlated; weights are researcher-defined; scores depend on the comparison set; missing-evidence penalties measure evidence completeness rather than observed poor uncertainty; and uncertainty methods are heterogeneous. Dimension-level evidence is primary. The composite is an exploratory secondary summary.
